```python
#Nested Loop
countK=0
counti=0
countj=0
for i in range(5):
    print("outer loop",i)
    counti+=1
    for j in range(5):
        print ("inner Loop",j)
        countj+=1
        for k in range(5):
            print('internal loop',k)
            countK+=1

print(countK,countj,counti)
```

```python
#Set Operations
#1.Union (| or union())- Combines all unique elements.
A = {1, 2, 3}
B = {3, 4, 5}
print(A | B)
```

```python
#2. Intersection (& or intersection())- Returns common elements.
A = {1, 2, 3}
B = {2, 3, 4}
print(A & B)
```

```python
#3. Difference (- or difference())- Returns elements present in the first set but not in the second.
A = {1, 2, 3}
B = {2, 3, 4}
print(A - B)
```

```python
#4. Symmetric Difference (^ or symmetric_difference())- Returns elements that are in either set but not in both.
A = {1, 2, 3}
B = {2, 3, 4}
print(A ^ B)
```

```python
#5. issubset()- Checks whether one set is a subset of another.
A = {1,2,3}
B = {1, 2, 3, 4,5}
print(A.issubset(B))
```

```python
#6. issuperset()- Checks whether one set is a superset of another.
#superset=A superset is a set that contains all the elements of another set (and it may contain additional elements as well).
A = {1, 2, 3, 4}
B = {1, 2}
print(A.issuperset(B))
```
