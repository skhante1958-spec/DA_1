```python
#*args- Arguments
#*args allows a function to accept any number of positional arguments.
#Instead of specifying a fixed number of parameters, Python collects all positional arguments into a tuple.
def numbers(*args):
    print(args)

numbers(30,"20","A",50,{"A":20.25})
```

```python
def total(*numbers):
    print(numbers)
    print(sum(numbers))

total(10,20)
total(10,20,30)
total(10,20,30,40)
```

```python
#**kwargs- Key value pair
#**kwargs allows a function to accept any number of keyword arguments.
#Python stores them inside a dictionary.
def student(**kwargs):
    print(kwargs)

student(name="John", age=22, city="Nagpur")
```

```python
#Difference: *args vs **kwargs
#*args	                           **kwargs
#Accepts positional arguments	     Accepts keyword arguments
#Stored as tuple	                 Stored as dictionary
#Uses single *	                     Uses double **
```

```python
#Combining Normal Arguments, *args and **kwargs

def student(name,last_name,*subjects, **details):
    print("Name :", name,last_name)
    print("Subjects :", subjects)
    print("Details :", details)

student("John","Doe",20,30,"Python","SQL", city="Nagpur",age=22)
```

```python
#Combining Normal Arguments, *args and **kwargs
def student(abc,name,name1="XYZ",*subjects,**details):
    print("abc :", abc)
    print("Name :",name,)
    print("name1 :",name1,)
    print("Subjects :", subjects)
    print("Details :", details)

student("John","Python","SQL",10,city="Nagpur",age=22)

#**details must always be the last parameter in a function definition.
#Once Python sees **details, it assumes all remaining keyword arguments belong to details. Therefore, no other parameter can come after it.
```

```python
#Lambda Functions
#A Lambda Function is an anonymous (nameless) function written in a single line.
#It is generally used for short operations.

#Normal Function	                           Lambda Function
#Uses def	                                    Uses lambda
#Can contain multiple statements	           Only one expression
#Has a function name	                        Anonymous
#Best for large logic	                    Best for short operations

square = lambda x: x*x

print(square(5))

add = lambda a,b : a+b

print(add(10,20))

largest = lambda a,b : a if a>b else b

print(largest(15,20))
```

```python
#Multiple except Blocks
number=0
try:
    number = int(input("Enter Number: "))
    number2=100
    result = number2/ number
    print(result)

except ValueError:
    print("Please enter a valid integer.")

except ZeroDivisionError:
    print("Number cannot be zero.",number2/2)
```

```python
#With Else
try:
    number = int(input("Enter Number: "))
    print(100 / number)

except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print("Program executed successfully.")
```

```python
#finally Block
#The finally block always executes whether an exception occurs or not.
#It is commonly used to close files, database connections, or release resources.
try:
    number = int(input("Enter Number: "))
    number2=100
    result = number2/ number
    print(result)

except ZeroDivisionError:
    print("Error")

finally:
    print("Program Finished finally")
```

```python
#else	                                          finally
#Executes only if no exception occurs	         Executes every time
#Optional	                                     Optional
#Used after successful execution	             Used for cleanup

#try → Always executes first.
#except → Executes only if an exception occurs and it matches.
#else → Executes only if no exception occurs.
#finally → Executes always, whether an exception occurs or not.
```
